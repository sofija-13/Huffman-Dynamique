```python

s = "é—漢"               # exemple avec caractères multi-octets
b = s.encode('utf-8')    # bytes UTF-8

ba = bitarray()
ba.frombytes(b)          # ajoute les bits correspondant aux bytes

print("bytes:", b)
print("bits :", ba.to01())          # chaîne '0'/'1' de tous les bits
print("per-octet:", ' '.join(format(x, '08b') for x in b))

# retransformer en chaîne
b2 = ba.tobytes()
s2 = b2.decode('utf-8')
print("decoded:", s2)
```
