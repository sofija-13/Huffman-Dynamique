```python
#ctor
root = Node(0)
createTree(root, 3)
print(root)
```

```python
# recherche pour une feuille dans l'arbre
print(root.searchNode(14, verbose=True))
# recherche pour un node dans l'arbre
print(root.searchNode(4, True, True))
# recherche de node non présent
print(root.searchNode(88, searchAll=True, verbose=True))
```

```python
# non feuille
print(root.isLeaf())
# feuille
print(root.searchNode(12).isLeaf())
# non node
print(root.searchNode(12).isNode())
# node
print(root.searchNode(5, searchAll=True).isNode())
# non root 
print(root.searchNode(5, searchAll=True).isRoot())
# root 
print(root.isRoot())
```

```python
# length arbre
print(root.length())
allLeft : Node = root.searchNode(7)
allLeft.newLeft(78)
print(root.length())
```

```python
# distance à la racine
print(root)
print(allLeft.distFromRoot()) # dist(7,0)
```

```python
# swap entre deux nodes
swapNodes(root.searchNode(8, True), root.searchNode(6, True), verbose=True)
print(root)
```

```python
# voisin gauche
print(root.searchNode(5, searchAll=True).getLeftNeighbor().value)
# voisin droit 
print(root.searchNode(5, searchAll=True).getRightNeighbor().value)
```
