#### "carambarbcm"

```python
# Compression
!echo "carambarbcm" > carambar.txt 
!../compresser carambar.txt carambar.huff
```

```python
# Résultat
with open("carambar.huff", 'rb') as f: 
    print(''.join(format(b, '08b') for b in f.read()))
```

```python
# Décompression
!../decompresser carambar.huff decodedcarambar.txt
!echo OK
!cat decodedcarambar.txt
```
